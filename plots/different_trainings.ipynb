```python
import pandas as pd

lst = [twenty, hundred, two_hundred, four, hezar, full]
lr = pd.DataFrame(columns=['Model', 'F1_macro', 'Accuracy','Precision','Recall','Specificity' ])
SVM = pd.DataFrame(columns=['Model', 'F1_macro', 'Accuracy','Precision','Recall','Specificity' ])
DT = pd.DataFrame(columns=['Model', 'F1_macro', 'Accuracy','Precision','Recall','Specificity' ])
knn = pd.DataFrame(columns=['Model', 'F1_macro', 'Accuracy','Precision','Recall','Specificity' ])
RF = pd.DataFrame(columns=['Model', 'F1_macro', 'Accuracy','Precision','Recall','Specificity' ])
XGboost = pd.DataFrame(columns=['Model', 'F1_macro', 'Accuracy','Precision','Recall','Specificity' ])
NN = pd.DataFrame(columns=['Model', 'F1_macro', 'Accuracy','Precision','Recall','Specificity' ])

for batch in lst:
    for model in batch['Model'].tolist():
        if model == 'logistic regression':
            lr = pd.concat([lr, batch.loc[batch['Model'] == f'{model}']])
        elif model == 'SVM':
            SVM = pd.concat([SVM, batch.loc[batch['Model'] == f'{model}']])
        elif model == 'Decision tree':
            DT = pd.concat([DT, batch.loc[batch['Model'] == f'{model}']])
        elif model == 'knn':
            knn = pd.concat([knn, batch.loc[batch['Model'] == f'{model}']])
        elif model == 'Random forest':
            RF = pd.concat([RF, batch.loc[batch['Model'] == f'{model}']])
        elif model == 'XGboost':
            XGboost = pd.concat([XGboost, batch.loc[batch['Model'] == f'{model}']])
        else:
            NN = pd.concat([NN, batch.loc[batch['Model'] == f'{model}']])
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, precision_score, recall_score, f1_score
from sklearn.metrics import roc_auc_score

def calculate_metrics(y_test, y_predicted):
    accuracy = accuracy_score(y_test, y_predicted)
    cm = confusion_matrix(y_test, y_predicted)
    precision = precision_score(y_test, y_predicted)
    recall = recall_score(y_test, y_predicted)
    specificity = cm[0, 0] / (cm[0, 0] + cm[0, 1]) if (cm[0, 0] + cm[0, 1]) != 0 else 0
    f1 = f1_score(y_test, y_predicted)
    
    return accuracy, precision, recall, specificity, f1
# def calculate_and_update_model_metrics(y_test, y_predicted):
#     accuracy, precision, recall, specificity, f1 = calculate_metrics(y_test, y_predicted)
#     model = {}
#     model["Accuracy"] = accuracy
#     model["Precision"] = precision
#     model["Recall"] = recall
#     model["Specificity"] = specificity
#     model["F1"] = f1
#    # model["AUC"] = roc_auc_score(y_test, y_predicted)
    
#     return model
# from sklearn.metrics import f1_score
# from scipy.stats import ttest_1samp
# from sklearn.model_selection import cross_val_score

def calculate_and_update_model_metrics(y_test, y_predicted):

    accuracy, precision, recall, specificity, f1 = calculate_metrics(y_test, y_predicted)
    f1_macro = f1_score(y_test, y_predicted, average='macro')


    model = {}
    model["Accuracy"] = accuracy
    model["precision"] = precision
    model["recall"] = recall
    model["specificity"] = specificity
    model["f1"] = f1
    model["F1_macro"] = f1_macro
    model["AUC"] = roc_auc_score(y_test, y_predicted)

    return model
```
